```python
from keras.datasets import imdb
import tensorflow as tf
from keras.preprocessing import sequence
```

```python
max_features = 10000
maxlen = 500
batch_size =32
```

```python
(input_train, y_train), (input_test, y_test) = imdb.load_data(num_words=max_features)
```

```python
print(len(input_train), "train sequences")
print(len(input_test), "test sequences")
```

```python
input_train = tf.keras.preprocessing.sequence.pad_sequences(input_train, maxlen=maxlen)
input_test = tf.keras.preprocessing.sequence.pad_sequences(input_test, maxlen=maxlen)
```

```python
print("input_train shape:", input_train.shape)
print("input_test shape:", input_test.shape)
```

```python
from keras.layers import LSTM, Dense, Embedding
from keras.models import Sequential
```

```python
model = Sequential()
```

```python
model.add(Embedding(max_features, 32))
model.add(LSTM(32))
model.add(Dense(1, activation="sigmoid"))
```

```python
model.summary()
```

```python
model.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["acc"]
)
```

```python
history = model.fit(
    input_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2
)
```

```python
acc = history.history["accuracy"]
val_acc = history.history["val_accuracy"]

loss = history.history["loss"]
val_loss = history.history["val_loss"]

epochs = range(1, len(acc) + 1)
```

```python
# Alternatif LSTM denemesi

from keras.datasets import imdb
from keras.preprocessing import sequence
import tensorflow as tf

max_features = 10000   # Dikkate alınacak kelime sayısı
maxlen = 500          # Metni 1000 kelimeden sonra keser
batch_size = 32
```

```python
(input_train, y_train), (input_test, y_test) = imdb.load_data(num_words=max_features)
```

```python
input_train = tf.keras.preprocessing.sequence.pad_sequences(input_train, maxlen=maxlen)
input_test = tf.keras.preprocessing.sequence.pad_sequences(input_test, maxlen=maxlen)
```

```python
print("input_train shape:", input_train.shape)
print("input_test shape:", input_test.shape)
```

```python
from keras.layers import LSTM, Dense, Embedding
from keras.models import Sequential

model = Sequential()

model.add(Embedding(max_features, 32))
model.add(LSTM(64))
model.add(Dense(1, activation="sigmoid"))

model.summary()
```

```python
model.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["acc"]
)
```

```python
history = model.fit(
    input_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)
```

```python
import matplotlib.pyplot as plt

acc = history.history["acc"]
val_acc = history.history["val_acc"]
loss = history.history["loss"]
val_loss = history.history["val_loss"]

epochs = range(1, len(acc) + 1)
```

```python
plt.plot(epochs, acc, "bo", label="Eğitim başarımı")
plt.plot(epochs, val_acc, "b", label="Doğrulama başarımı")
plt.title("Eğitim ve doğrulama başarımı")
plt.legend()

plt.show()
plt.clf()
```

```python
plt.plot(epochs, loss, "bo", label="Eğitim kaybı")
plt.plot(epochs, val_loss, "b", label="Doğrulama kaybı")
plt.title("Eğitim ve doğrulama kaybı")
plt.legend()

plt.show()
```
