```python
from tensorflow.keras import layers, models

model = models.Sequential()

model.add(layers.Input(shape=(150, 150, 3)))

model.add(layers.Conv2D(32, (3, 3), activation="relu"))
model.add(layers.MaxPooling2D((2, 2)))

model.add(layers.Conv2D(64, (3, 3), activation="relu"))
model.add(layers.MaxPooling2D((2, 2)))

model.add(layers.Conv2D(128, (3, 3), activation="relu"))
model.add(layers.MaxPooling2D((2, 2)))

model.add(layers.Conv2D(128, (3, 3), activation="relu"))
model.add(layers.MaxPooling2D((2, 2)))

model.add(layers.Flatten())
model.add(layers.Dense(512, activation="relu"))
model.add(layers.Dense(1, activation="sigmoid"))

model.summary()
```

```python
from keras import optimizers
```

```python
model.compile(
    loss="binary_crossentropy",
    optimizer=optimizers.RMSprop(learning_rate=1e-4),
    metrics=["accuracy"]
)
```

```python
train = ImageDataGenerator(rescale=1./255)
validation = ImageDataGenerator(rescale=1./255)
test = ImageDataGenerator(rescale=1./255)
```

```python
train_dataset = train.flow_from_directory(
    r'C:\Users\User\Desktop\cats_and_dogs_small\train',  # Added 'r' prefix for raw string
    target_size=(150, 150),
    batch_size=20,
    class_mode="binary"
)

validation_datasets = validation.flow_from_directory(
    r'C:\Users\User\Desktop\cats_and_dogs_small\validations',  # Added 'r' prefix for raw string
    target_size=(150, 150),
    batch_size=20,
    class_mode="binary"
)
```

```python
history = model.fit(
    train_dataset,
    steps_per_epoch=100,
    epochs=30,
    validation_data=validation_datasets,
    validation_steps=50
)
```

```python
model.save("cats_and_dogs_small.h5")
```

```python
acc = history.history["accuracy"]
val_acc = history.history["val_accuracy"]

loss = history.history["loss"]
val_loss = history.history["val_loss"]

epochs = range(1, len(acc) + 1)

plt.plot(epochs, acc, "bo", label="Eğitim başarımı")
plt.plot(epochs, val_acc, "b", label="Doğrulama başarımı")
plt.title("Eğitim ve doğrulama başarımı")
plt.legend()
plt.show()
```

```python
plt.plot(epochs, loss, "bo", label="Eğitim kaybı")
plt.plot(epochs, val_loss, "b", label="Doğrulama kaybı")
plt.title("Eğitim ve doğrulama kaybı")
plt.legend()
plt.show()
```
